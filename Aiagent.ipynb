```python
from langchain.agents import create_agent

from langchain_groq import ChatGroq

import os
from dotenv import load_dotenv
load_dotenv()
groq_api_key = os.getenv("GROQ_API_KEY")

llm= ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0.5, 
    max_retries=2,
    api_key=groq_api_key
)

import requests
def get_weather(city: str) -> str:
    """Get weather for a given city."""

    url = "https://p2pclouds.up.railway.app/v1/learn/weather"
    
    try:
        response = requests.get(url, params={"city": city}, timeout=10)
        response.raise_for_status()

        data = response.json()

        # Return the complete response as a string
        return str(data)

    except requests.RequestException as e:
        return f"Error fetching weather: {e}"

    
def get_my_name() -> str:
    """Get the AI agent name."""
    return "Niki"

#Library to save the Chat in memory.
from langgraph.checkpoint.memory import InMemorySaver
#library is available but we also need to access the Memory so we need a instance to get access.
memory= InMemorySaver()

agent = create_agent(
    model=llm,
    tools=[get_weather, get_my_name],
    system_prompt="You are a helpful agent",
    checkpointer=memory

)
```

```python
response=agent.invoke(
    {"messages": [
        {
            "role": "user",
            "content": "How persistent is Your Memory?"
            }]},
            config={"configurable":{"thread_id":"12345678"}}
)
print(response['messages'][-1])
```
